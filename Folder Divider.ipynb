```python
import os
import shutil

# Path to the folder containing all images
source_folder = r'C:\Users\snicolas\archive (1)\Male and Female face dataset\Faces'

# Path to the destination folder where subfolders will be created
destination_folder = r'C:\Users\snicolas\archive (1)\Male and Female face dataset\Divided_Folders'

# Number of images per folder
images_per_folder = 100

# Create the destination folder if it doesn't exist
os.makedirs(destination_folder, exist_ok=True)

# Get a list of all image files in the source folder
image_files = [f for f in os.listdir(source_folder) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

# Divide images into smaller subfolders
folder_index = 1
for i in range(0, len(image_files), images_per_folder):
    # Create a new subfolder
    subfolder_path = os.path.join(destination_folder, f'Folder_{folder_index}')
    os.makedirs(subfolder_path, exist_ok=True)
    
    # Move or copy the current batch of images into the subfolder
    for image_file in image_files[i:i + images_per_folder]:
        source_path = os.path.join(source_folder, image_file)
        destination_path = os.path.join(subfolder_path, image_file)
        shutil.move(source_path, destination_path)  # Use shutil.copy if you want to copy instead of move
    
    print(f'Created {subfolder_path} with {len(image_files[i:i + images_per_folder])} images.')
    folder_index += 1

print("Image division completed!")
```
