```python
from cassandra.cluster import Cluster
from cassandra.query import PreparedStatement
import os

# Connect to Cassandra
cluster = Cluster(['172.31.13.93'])  # Replace with your Cassandra host
session = cluster.connect()

# Use the keyspace where the 'images' table is created
session.set_keyspace('datacrowd')  # Replace with your keyspace name

# Prepare the insert statement
insert_image = session.prepare("""
    INSERT INTO images (image_id, image_name, image, gender) 
    VALUES (?, ?, ?, ?)
""")

# Path to your image directory
image_directory = 'C:\Users\snicolas\archive (1)\Male and Female face dataset\Female Faces'  # Replace with the path to your images folder

# Function to insert images
def insert_images():
    image_id = 1  # Initialize image_id
    for filename in os.listdir(image_directory):
        if filename.lower().endswith(('.jpg', '.jpeg', '.png')):
            with open(os.path.join(image_directory, filename), 'rb') as img_file:
                binary_data = img_file.read()
                # Insert image into the Cassandra table
                session.execute(insert_image, (image_id, filename, binary_data, "unknown"))
                print(f"Inserted: {filename} with ID {image_id}")
                image_id += 1

# Run the function
insert_images()

# Close the connection
cluster.shutdown()
```
